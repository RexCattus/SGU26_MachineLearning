# Lab 03: Khám phá Dữ liệu (EDA) - Ames Housing
**Mục tiêu:** Phân tích tập dữ liệu Ames Housing để xác định tính chất, xử lý giá trị thiếu, loại bỏ ngoại lệ và khảo sát tương quan đa biến trước khi xây dựng mô hình Hồi quy đa biến.

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
pd.set_option('display.max_columns', 100)

# Đọc dữ liệu trên Kaggle
train_file = '/kaggle/input/house-prices-advanced-regression-techniques/train.csv'
if not os.path.exists(train_file):
    # Đường dẫn dự phòng nếu thư mục input có tên khác
    train_file = '/kaggle/input/competitions/house-prices-advanced-regression-techniques/train.csv'

train = pd.read_csv(train_file)
print(f"Kích thước tập dữ liệu ban đầu: {train.shape}")
```

## 1. Phân tích Đơn biến: Biến mục tiêu (SalePrice)
Khảo sát phân phối của giá bán nhà. Việc hiểu rõ biến mục tiêu giúp xác định xem có cần thực hiện các phép biến đổi toán học (như logarit) để ổn định phương sai hay không.

```python
plt.figure(figsize=(10, 6))
sns.histplot(train['SalePrice'], kde=True, color='royalblue', bins=50)
plt.title('Phan phoi bien muc tieu SalePrice')
plt.xlabel('Giá bán (USD)')
plt.ylabel('Tần suất')
plt.show()

print(f"Độ lệch (Skewness): {train['SalePrice'].skew():.4f}")
print(f"Độ nhọn (Kurtosis): {train['SalePrice'].kurt():.4f}")
```

## 2. Xác định và Xử lý Giá trị thiếu (Missing Values)
Phân tích nguyên nhân thiếu hụt dữ liệu để có chiến lược điền giá trị phù hợp, thay vì điền trung bình/mode một cách máy móc cho toàn bộ tập dữ liệu. Nhiều biến bị `NaN` thực chất mang ý nghĩa là bất động sản đó không sở hữu hạng mục tương ứng (ví dụ: không có hồ bơi, không có lò sưởi).

```python
# Tính tỷ lệ phần trăm dữ liệu thiếu
missing_data = train.isnull().sum()
missing_percent = (missing_data[missing_data > 0] / len(train)) * 100
missing_percent = missing_percent.sort_values(ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=missing_percent.index, y=missing_percent.values, palette='viridis')
plt.xticks(rotation=90)
plt.title('Ty le % du lieu thieu theo tung dac trung')
plt.ylabel('%')
plt.show()

# Xử lý thiếu hụt có chủ đích
# Điền 'None' cho các hạng mục mà NaN có nghĩa là không tồn tại
none_cols = ['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'FireplaceQu', 
             'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond',
             'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2', 'MasVnrType']
for col in none_cols:
    train[col] = train[col].fillna('None')

# Điền 0 cho các biến định lượng liên quan đến diện tích/số lượng khi hạng mục không tồn tại
zero_cols = ['GarageArea', 'GarageCars', 'BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 'BsmtFullBath', 'BsmtHalfBath', 'MasVnrArea']
for col in zero_cols:
    train[col] = train[col].fillna(0)

# Các biến phân loại thực sự bị thiếu, điền bằng mode
mode_cols = ['Electrical']
for col in mode_cols:
    train[col] = train[col].fillna(train[col].mode()[0])
    
# Cột LotFrontage (Mặt tiền) có thể điền bằng trung vị của Neighborhood tương ứng
train['LotFrontage'] = train.groupby('Neighborhood')['LotFrontage'].transform(lambda x: x.fillna(x.median()))

print("Số lượng giá trị thiếu còn lại:", train.isnull().sum().max())
```

## 3. Xác định và Loại bỏ Ngoại lệ (Outliers)
Theo tác giả De Cock, để mô hình hoạt động ổn định, cần loại bỏ các quan sát có diện tích không gian sống trên mặt đất (`GrLivArea`) lớn hơn 4000 feet. Các điểm này đại diện cho các giao dịch bán một phần hoặc các ngôi nhà lớn bất thường không phản ánh đúng giá trị thị trường thực tế.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=train['GrLivArea'], y=train['SalePrice'], color='darkorange', alpha=0.7)
plt.axvline(x=4000, color='red', linestyle='--', label='Nguong ngoai lai (> 4000 sq ft)')
plt.title('Phan tich ngoai le: GrLivArea vs SalePrice')
plt.xlabel('Dien tich sinh hoat tren mat dat (sq ft)')
plt.ylabel('Gia ban (USD)')
plt.legend()
plt.show()

# Thực hiện loại bỏ ngoại lệ
train = train[train['GrLivArea'] <= 4000]
print(f"Kích thước tập dữ liệu sau khi loại bỏ ngoại lệ: {train.shape}")
```

## 4. Trích xuất Đặc trưng (Feature Engineering) và Phân tích Đa biến
Theo tài liệu nghiên cứu, một mô hình hồi quy đơn giản có thể dự đoán tốt giá nhà dựa trên tổng diện tích mét vuông (bao gồm diện tích tầng hầm và diện tích mặt đất). Cần tạo đặc trưng `TotalSF` và khảo sát ma trận tương quan để định lượng mức độ ảnh hưởng tuyến tính của các biến số học đến giá bán.

```python
# Tạo biến tổng diện tích
train['TotalSF'] = train['TotalBsmtSF'] + train['GrLivArea']

# Chọn các biến định lượng để tính tương quan
numeric_cols = train.select_dtypes(include=['int64', 'float64']).columns
corr_matrix = train[numeric_cols].corr()

# Lấy top 10 biến có độ tương quan cao nhất với SalePrice
top_corr_features = corr_matrix.nlargest(11, 'SalePrice').index

plt.figure(figsize=(10, 8))
sns.heatmap(train[top_corr_features].corr(), annot=True, cmap='coolwarm', fmt=".2f", square=True)
plt.title('Ma tran tuong quan đa bien (Top 10 đac trung)')
plt.show()

# Trực quan hóa tương quan tuyến tính của đặc trưng mới tạo
plt.figure(figsize=(10, 6))
sns.regplot(x=train['TotalSF'], y=train['SalePrice'], scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.title('Quan he tuyen tinh giua tong dien tich (TotalSF) và Gia ban')
plt.xlabel('Tong dien tich (sq ft)')
plt.ylabel('Gia ban (USD)')
plt.show()
```
