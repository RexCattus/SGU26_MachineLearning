## Bài 8. Autograd cơ bản

```python
import torch
x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print('x.grad của y = x^2 tại x=3.6 là:', x.grad.item())

# Đổi hàm thành y = 3x^2 + 2x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2**2 + 2 * x2 + 1
y2.backward()
print('x.grad của y = 3x^2 + 2x + 1 tại x=3.6 là:', x2.grad.item())
```

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 50
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([X * X, X, np.ones_like(X)])

w = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX, dtype=torch.float32)
y_tensor = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

print('Hệ số học được:\n', w.detach().numpy())
```

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print(f'Nghiệm thu được: A={A.item():.2f}, B={B.item():.2f}, C={C.item():.2f}, D={D.item():.2f}')
print(f'Kiểm tra:\n A+B={(A+B).item():.2f} (Kỳ vọng: 9)\n C-D={(C-D).item():.2f} (Kỳ vọng: 1)\n A+C={(A+C).item():.2f} (Kỳ vọng: 8)\n B-D={(B-D).item():.2f} (Kỳ vọng: 2)')
```

```python
# 3. Tính dy/dx của y = 2x^2 + 5x + 3 tại x = 2
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print('Gradient (dy/dx) tại x = 2 bằng PyTorch:', x.grad.item())
print('Kiểm tra bằng tính thủ công: y\' = 4x + 5. Tại x=2, y\' = 4(2) + 5 = 13')
```

```python
# 4. Thay đa thức bằng y = 2x^2 - 3x + 5 và tìm hệ số
import numpy as np
polynomial = np.poly1d([2, -3, 5])
N = 100
X_data = np.random.randn(N, 1) * 5
Y_data = polynomial(X_data)
XX_data = np.hstack([X_data * X_data, X_data, np.ones_like(X_data)])

w = torch.randn(3, 1, requires_grad=True)
x_tensor = torch.tensor(XX_data, dtype=torch.float32)
y_tensor = torch.tensor(Y_data, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.05)

for epoch in range(3000):
    optimizer.zero_grad()
    y_pred = x_tensor @ w
    mse = torch.mean(torch.square(y_tensor - y_pred))
    mse.backward()
    optimizer.step()

print('Hệ số học được (Kỳ vọng: [2, -3, 5]):\n', w.detach().numpy())
```
